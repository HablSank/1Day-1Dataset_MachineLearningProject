# K-Nearest Neighbors (Bangladesh Diabetes Prediction)

## Penjelasan

### Tujuan

Untuk membuat model machine learning yang dapat memprediksi apakah suatu pasien menderita diabetes atau tidak melalui data yang telah disediakan.

### Penjelasan Kolom

- ***Age (Usia)*** = Usia pasien dalam tahun.
- ***Gender (Jenis Kelamin)*** = Jenis kelamin pasien (Pria/Wanita).
- ***Polyuria*** = Frekuensi buang air kecil yang berlebihan (Ya/Tidak).
- ***Polydipsia*** = Rasa haus yang berlebihan (Ya/Tidak).
- ***Sudden weight loss (Penurunan berat badan mendadak)*** = Penurunan berat badan yang tiba-tiba tanpa sebab jelas
(Ya/Tidak).
- ***Weakness (Kelemahan)*** = Perasaan lemah atau lelah yang tidak biasa (Ya/Tidak).
- ***Polyphagia*** = Nafsu makan yang meningkat secara berlebihan (Ya/Tidak).
- ***Genital thrush (Infeksi jamur genital)*** = Adanya infeksi jamur pada area genital (Ya/Tidak).
- ***Visual blurring (Penglihatan kabur)*** = Gangguan penglihatan seperti penglihatan kabur (Ya/Tidak).
- ***Itching (Gatal-gatal)*** = Rasa gatal pada kulit tanpa sebab yang jelas (Ya/Tidak).
- ***Irritability (Iritabilitas)*** = Perasaan mudah marah atau tersinggung (Ya/Tidak).
- ***Delayed healing (Penyembuhan luka lambat)*** = Luka yang membutuhkan waktu lebih lama untuk sembuh (Ya/Tidak).
- ***Partial paresis (Paresis parsial)*** = Kelemahan otot sebagian atau kelumpuhan ringan (Ya/Tidak).
- ***Muscle stiffness (Kekakuan otot)*** = Kekakuan atau ketegangan pada otot (Ya/Tidak).
- ***Alopecia*** = Kehilangan rambut atau kebotakan (Ya/Tidak).
- ***Obesity (Obesitas)*** = Kondisi kelebihan berat badan atau obesitas (Ya/Tidak).
- ***Class*** = Klasifikasi apakah pasien menderita diabetes atau tidak (Positif/Negatif).
